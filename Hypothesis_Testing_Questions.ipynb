#Q1)What is a null hypothesis (H₀) and why is it important in hypothesis testing?
#Ans-->

### --> What is a Null Hypothesis (H₀)?
- The **null hypothesis (H₀)** is a formal statement in statistics that assumes **no effect, no difference, or no relationship** exists between variables.
- It represents the **default or baseline assumption** before testing.  
  Example: If you’re testing whether a new drug lowers blood pressure, the null hypothesis would be:  
  *“The drug has no effect on blood pressure compared to the placebo.”*

---

### --> Why is it Important?
1. **Provides a Starting Point**  
   - Hypothesis testing begins with H₀ as the foundation. Without it, we wouldn’t have a clear benchmark to test against.

2. **Ensures Objectivity**  
   - By assuming no effect initially, researchers avoid bias and let the data speak for itself.

3. **Basis for Statistical Tests**  
   - Most statistical methods (t-tests, chi-square tests, ANOVA, regression) are designed to evaluate whether we have enough evidence to reject H₀.

4. **Controls Error Rates**  
   - Testing against H₀ helps manage **Type I errors** (false positives). We only reject H₀ when evidence is strong enough (usually at a significance level like α = 0.05).

5. **Scientific Rigor**  
   - It enforces discipline: extraordinary claims require extraordinary evidence. Rejecting H₀ means we’ve found statistically significant results worth further investigation.

---

### --> Quick Analogy
Think of H₀ as the **“innocent until proven guilty”** principle in law:
- H₀ = The defendant is innocent.  
- Evidence (data) must be strong enough to reject H₀.  
- If evidence is weak, we fail to reject H₀ (we don’t prove innocence, we just don’t have proof of guilt).

#Q2) What does the significance level (α) represent in hypothesis testing?
#Ans-->
### --> Significance Level (α) in Hypothesis Testing
- The **significance level (α)** is the **threshold probability** used to decide whether to reject the null hypothesis (H₀).
- It represents the **maximum risk of making a Type I error** (rejecting H₀ when it is actually true).

---

### --> What Does It Mean?
- If you set **α = 0.05**, you are saying:  
  *“I am willing to accept a 5% chance of incorrectly rejecting the null hypothesis.”*
- In other words, α defines how strong the evidence against H₀ must be before you conclude that the result is statistically significant.

---

### --> Common Values of α
- **0.05 (5%)** → Most widely used in scientific studies.  
- **0.01 (1%)** → Stricter standard, used when consequences of error are serious (e.g., medical trials).  
- **0.10 (10%)** → More lenient, sometimes used in exploratory research.

---

### --> Role in Hypothesis Testing
1. **Comparison with p-value**  
   - If *p-value ≤ α*: Reject H₀ (evidence is strong enough).  
   - If *p-value > α*: Fail to reject H₀ (not enough evidence).

2. **Controls False Positives**  
   - α sets the boundary for how cautious or bold you are in declaring results significant.

3. **Balances Risk vs. Discovery**  
   - Lower α → More cautious, fewer false positives, but higher chance of missing real effects (Type II error).  
   - Higher α → More discoveries, but greater risk of false positives.

---

### --> Quick Analogy
Think of α as the **security threshold at an airport**:
- A strict threshold (α = 0.01) means only very suspicious signals trigger alarms (fewer false alarms, but some threats may slip through).  
- A lenient threshold (α = 0.10) means more alarms go off (catching more threats, but also more false alarms).

#Q3)  Differentiate between Type I and Type II errors.
#Ans-->


### --> Type I Error (False Positive)
- **Definition:** Rejecting the null hypothesis (H₀) when it is actually true.  
- **Analogy:** Convicting an innocent person in court.  
- **Risk:** Controlled by the **significance level (α)**.  
- **Example:** A medical test says a patient has a disease when they actually don’t.  

---

### --> Type II Error (False Negative)
- **Definition:** Failing to reject the null hypothesis (H₀) when it is actually false.  
- **Analogy:** Letting a guilty person go free.  
- **Risk:** Represented by **β**, the probability of a Type II error.  
- **Power of a Test:** Defined as \(1 - \beta\), which measures the ability to correctly detect a true effect.  
- **Example:** A medical test says a patient is healthy when they actually have the disease.  

---

### --> Key Differences

| Aspect                | Type I Error (α) | Type II Error (β) |
|-----------------------|------------------|-------------------|
| What happens?         | Reject H₀ when true | Fail to reject H₀ when false |
| Nickname              | False Positive   | False Negative    |
| Controlled by         | Significance level (α) | Test power (1 - β) |
| Analogy               | Innocent convicted | Guilty acquitted |
| Example               | Healthy person diagnosed sick | Sick person diagnosed healthy |

---

### --> Why It Matters
- **Type I error** → We claim an effect exists when it doesn’t (risk of false discoveries).  
- **Type II error** → We miss a real effect (risk of overlooking important findings).  
- Researchers must balance these risks depending on context:  
  - In medicine, avoiding **Type I errors** (false positives) is crucial to prevent unnecessary treatments.  
  - In safety-critical fields, avoiding **Type II errors** (false negatives) is vital to catch real dangers.  



#Q4)  Explain the difference between a one-tailed and two-tailed test. Give an example of each.
#Ans-->

### --> One-Tailed Test
- **Definition:** A one-tailed test checks for an effect in **only one direction** (greater than OR less than).
- **When to use:** If your research question is directional.  
  Example: *“Does the new drug lower blood pressure compared to placebo?”*  
  - **H₀:** The drug has no effect (mean difference = 0).  
  - **H₁ (one-tailed):** The drug lowers blood pressure (mean difference < 0).  
- **Interpretation:** You only look at one end of the distribution (left tail or right tail).

---

### --> Two-Tailed Test
- **Definition:** A two-tailed test checks for an effect in **both directions** (greater OR less).
- **When to use:** If your research question is non-directional.  
  Example: *“Does the new drug affect blood pressure compared to placebo?”*  
  - **H₀:** The drug has no effect (mean difference = 0).  
  - **H₁ (two-tailed):** The drug changes blood pressure (mean difference ≠ 0).  
- **Interpretation:** You split the significance level (α) across both tails of the distribution.

---

### --> Key Differences

| Aspect              | One-Tailed Test | Two-Tailed Test |
|---------------------|-----------------|-----------------|
| Direction of effect | One direction only | Both directions |
| Sensitivity         | More powerful for detecting effect in specified direction | More conservative, requires stronger evidence |
| Example             | “Drug lowers BP” | “Drug affects BP (increase or decrease)” |

---

### --> Quick Analogy
Think of tossing a coin:
- **One-tailed test:** You only care if it lands on **heads** more often than expected.  
- **Two-tailed test:** You care if it lands on **either heads or tails** more often than expected.


#Q5) : A company claims that the average time to resolve a customer complaint is 10 minutes.
A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes.
At α = 0.05, test the claim.
#Ans-->
**Performing a t-test for the mean**

I’m testing a hypothesis about the mean with a small sample size (n=9) and unknown sigma, so I’ll use a t-test. The claim is that the average is 10, so H0: μ=10. The sample mean is 12, with s=3. I calculate t = (12-10)/(3/√9) = 2. The degrees of freedom (df) is 8. Since it's a two-tailed test, I compare the calculated t-value to the critical t-value (≈2.306). Since 2.0 < 2.306, I fail to reject H0. The p-value is around 0.078, and the 95% confidence interval includes 10, supporting the claim.
### Hypothesis test for the mean (t-test)

**Given:** n=9, bar{x}=12minutes, s=3minutes, claimed mean mu=10 minutes, alpha=0.05.

#### Set up hypotheses
- **Null:** H_0:mu=10
- **Alternative:** H_1:mu !=10 (two-tailed, testing equality)

#### Test statistic
Compute the t-statistic using the sample standard deviation:
t=bar{x}-mu/s/sqrt(n)=12-10/3/sqrt(9)=2/1=2.0
Degrees of freedom: df=9-1=8

#### Critical value and decision
For a two-tailed test at alpha=0.05 with df=8, the critical value is since |t|=2.0 < 2.306, **fail to reject \(H_0\)**.

#### p-value (for context)
Two-tailed p-value for t=2.0 with df=8 is approximately p approx 0.078\), which is greater than \(0.05\).

---

### Conclusion
There isn’t sufficient evidence at the 5% significance level to reject the company’s claim that the average complaint resolution time is 10 minutes.



#Q6) When should you use a Z-test instead of a t-test?
#Ans-->


### --> Use a **Z-test** when:
- The **population standard deviation (sigma) is known**.
- The sample size is **large** (commonly \(n >= 30\), so the Central Limit Theorem ensures the sampling distribution of the mean is approximately normal.
- The population is normally distributed (especially important if \(n\) is small).

---

### --> Use a **t-test** when:
- The **population standard deviation is unknown** (you only have the sample standard deviation \(s\)).
- The sample size is **small** (\(n < 30\)), and you assume the population is approximately normal.
- You want to account for the extra uncertainty introduced by estimating \(sigma\) with \(s\).

---

### --> Key difference
- **Z-test** uses the **standard normal distribution** (mean 0, variance 1).
- **t-test** uses the **Student’s t-distribution**, which has heavier tails to reflect the added uncertainty from estimating \(\sigma\). As \(n\) grows large, the t-distribution approaches the normal distribution, so the distinction fades.

---

### Example
- If a company knows the **true population standard deviation** of complaint resolution times (say \(\sigma=3\) minutes), and takes a large sample, a **Z-test** is appropriate.  
- If the company only has a **sample standard deviation** from a small sample (like your earlier example with \(n=9\)), a **t-test** is the correct choice.

| Gender | Product A | Product B |

chi^2 = sum (O-E)^2/E

- Female, A: \(10-20)^2 / 20 = 5

chi^2 approx 16.67


---

### Decision

- Degrees of freedom: (2-(2-1) = 1\)  
- Critical value at alpha=0.05\: 3.841
- Since 16.67 > 3.841, we **reject the null hypothesis of independence**.

--> **Conclusion:** Product preference is **dependent on gender**. Male and female customers show significantly different preferences between Product A and Product B.
